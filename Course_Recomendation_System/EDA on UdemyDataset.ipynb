# EDA on Udemy Dataset

This notebook performs exploratory data analysis on the Udemy course dataset to understand the data patterns and insights.

```python
# Import libraries for EDA
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

# Set style for plots
plt.style.use('seaborn-v0_8')
sns.set_palette("husl")
```

```python
# Load the dataset
df = pd.read_csv('udemy_course_data.csv')
print(f"Dataset shape: {df.shape}")
print("\nDataset Info:")
df.info()
```

```python
# Display basic statistics
print("Basic Statistics:")
print(df.describe())

print("\nMissing Values:")
print(df.isnull().sum())
```

```python
# Check unique values in categorical columns
print("Unique subjects:")
print(df['subject'].value_counts())

print("\nUnique levels:")
print(df['level'].value_counts())
```

```python
# Data visualization - Subject distribution
plt.figure(figsize=(12, 6))
subject_counts = df['subject'].value_counts()
plt.subplot(1, 2, 1)
subject_counts.plot(kind='bar')
plt.title('Number of Courses by Subject')
plt.xticks(rotation=45)
plt.ylabel('Count')

plt.subplot(1, 2, 2)
subject_counts.plot(kind='pie', autopct='%1.1f%%')
plt.title('Course Distribution by Subject')
plt.ylabel('')

plt.tight_layout()
plt.show()
```

```python
# Level distribution
plt.figure(figsize=(10, 6))
level_counts = df['level'].value_counts()
plt.subplot(1, 2, 1)
level_counts.plot(kind='bar', color='skyblue')
plt.title('Number of Courses by Level')
plt.xticks(rotation=45)
plt.ylabel('Count')

plt.subplot(1, 2, 2)
level_counts.plot(kind='pie', autopct='%1.1f%%', colors=['lightcoral', 'lightblue', 'lightgreen'])
plt.title('Course Distribution by Level')
plt.ylabel('')

plt.tight_layout()
plt.show()
```

```python
# Price analysis
# Convert price to numeric, handling 'Free' and other text
df['price_numeric'] = pd.to_numeric(df['price'].replace(['Free', 'TRUE'], '0'), errors='coerce')

plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.hist(df['price_numeric'].dropna(), bins=30, alpha=0.7)
plt.title('Price Distribution')
plt.xlabel('Price ($)')
plt.ylabel('Frequency')

plt.subplot(1, 3, 2)
df.boxplot(column='price_numeric')
plt.title('Price Boxplot')
plt.ylabel('Price ($)')

plt.subplot(1, 3, 3)
price_by_subject = df.groupby('subject')['price_numeric'].mean().sort_values(ascending=False)
price_by_subject.plot(kind='bar')
plt.title('Average Price by Subject')
plt.xticks(rotation=45)
plt.ylabel('Average Price ($)')

plt.tight_layout()
plt.show()
```

```python
# Subscriber analysis
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.hist(df['num_subscribers'], bins=30, alpha=0.7, color='green')
plt.title('Subscriber Distribution')
plt.xlabel('Number of Subscribers')
plt.ylabel('Frequency')

plt.subplot(1, 3, 2)
subscribers_by_subject = df.groupby('subject')['num_subscribers'].sum().sort_values(ascending=False)
subscribers_by_subject.plot(kind='bar', color='orange')
plt.title('Total Subscribers by Subject')
plt.xticks(rotation=45)
plt.ylabel('Total Subscribers')

plt.subplot(1, 3, 3)
avg_subscribers = df.groupby('subject')['num_subscribers'].mean().sort_values(ascending=False)
avg_subscribers.plot(kind='bar', color='purple')
plt.title('Average Subscribers per Course by Subject')
plt.xticks(rotation=45)
plt.ylabel('Average Subscribers')

plt.tight_layout()
plt.show()
```

```python
# Correlation analysis
numeric_cols = ['price_numeric', 'num_subscribers', 'num_reviews', 'num_lectures']
correlation_matrix = df[numeric_cols].corr()

plt.figure(figsize=(10, 8))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', center=0)
plt.title('Correlation Matrix of Numeric Features')
plt.show()
```

```python
# Top courses by different metrics
print("Top 10 courses by subscribers:")
top_subscribers = df.nlargest(10, 'num_subscribers')[['course_title', 'subject', 'num_subscribers', 'price']]
print(top_subscribers)

print("\nTop 10 courses by reviews:")
top_reviews = df.nlargest(10, 'num_reviews')[['course_title', 'subject', 'num_reviews', 'price']]
print(top_reviews)
```

```python
# Time-based analysis if published_timestamp exists
if 'published_timestamp' in df.columns:
    # Convert to datetime
    df['published_date'] = pd.to_datetime(df['published_timestamp']).dt.date
    df['year'] = pd.to_datetime(df['published_timestamp']).dt.year
    
    plt.figure(figsize=(15, 5))
    
    plt.subplot(1, 2, 1)
    courses_by_year = df['year'].value_counts().sort_index()
    courses_by_year.plot(kind='bar')
    plt.title('Number of Courses Published by Year')
    plt.xlabel('Year')
    plt.ylabel('Number of Courses')
    
    plt.subplot(1, 2, 2)
    subscribers_by_year = df.groupby('year')['num_subscribers'].sum()
    subscribers_by_year.plot(kind='bar', color='red')
    plt.title('Total Subscribers by Publication Year')
    plt.xlabel('Year')
    plt.ylabel('Total Subscribers')
    
    plt.tight_layout()
    plt.show()
else:
    print("No timestamp column found for time-based analysis")
```

```python
# Data cleaning insights
print("Data Quality Report:")
print(f"Total courses: {len(df)}")
print(f"Courses with missing data: {df.isnull().any(axis=1).sum()}")
print(f"Duplicate courses: {df.duplicated().sum()}")

# Check for outliers in key metrics
print("\nPotential Outliers:")
print(f"Courses with >100k subscribers: {(df['num_subscribers'] > 100000).sum()}")
print(f"Courses with price >$200: {(df['price_numeric'] > 200).sum()}")
print(f"Courses with 0 subscribers: {(df['num_subscribers'] == 0).sum()}")
```
